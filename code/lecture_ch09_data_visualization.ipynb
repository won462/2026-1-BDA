```python
import matplotlib.pyplot as plt

# 데이터 준비
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

# 마커 종류별로 그래프 그리기
markers = ['.', ',', 'o', 'v', '^', '>', '<', 's', 'D', 'd', 'h', 'H', '*', '+', 'x', '|', '_']
labels = [
    "Point", "Pixel", "Circle", "Triangle_down", "Triangle_up", "Triangle_right",
    "Triangle_left", "Square", "Diamond", "Thin diamond", "Hexagon1", "Hexagon2",
    "Star", "Plus", "X", "Thin plus", "Thin x"
]

plt.figure(figsize=(10, 8))

for i, marker in enumerate(markers):
    plt.plot(x, [y_val + i for y_val in y], marker=marker, linestyle='None',
                                            label=labels[i], markersize=8)

# 그래프 제목과 축 레이블
plt.title("Marker Styles in Matplotlib")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()

# 그래프 보여주기
plt.show()
```

```python
# mask 만들기
import numpy as np
mask = np.zeros_like(corr_iris)
mask
```

```python
np.triu_indices_from(mask)
```

```python
# 오른쪽 위 대각 행렬을 1로 바꾸기
mask[np.triu_indices_from(mask)]
```
